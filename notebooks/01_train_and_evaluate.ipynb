# Personalized Healthcare ML — Training & Evaluation

This notebook mirrors `scripts/train_all.py` for classroom demonstration.

```python
from pathlib import Path
import sys, pandas as pd
ROOT = Path.cwd().resolve().parent
sys.path.insert(0, str(ROOT / 'src'))
from medireco.preprocess import load_dataset
from medireco.train import build_pipeline
```

```python
df = load_dataset(ROOT / 'data/raw/Cleaned_Dataset.csv')
df.head()
```

```python
X = df[['fever','cough','fatigue','difficulty_breathing','age','gender','blood_pressure','cholesterol_level']]
y = df['disease_group']
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report
model = build_pipeline()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.25, stratify=y, random_state=42)
model.fit(X_train, y_train)
pred = model.predict(X_test)
print('accuracy:', accuracy_score(y_test, pred))
print('balanced accuracy:', balanced_accuracy_score(y_test, pred))
print(classification_report(y_test, pred, zero_division=0))
```
